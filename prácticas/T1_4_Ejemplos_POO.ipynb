```python
# COMPOSICIÓN
class Salario:
    def __init__(self,pago):
        self.pago=pago

    def get_total(self):
       return (self.pago*12)

class Empleado:
    def __init__(self,pago,bonus):
        self.pago=pago
        self.bonus=bonus
        self.obj_salario=Salario(self.pago)

    def salario_anual(self):
        return "Total: "  +  str(self.obj_salario.get_total()+self.bonus)


obj_emp=Empleado(100,10)
print (obj_emp.salario_anual())
```

**RELACIONES UNO A MUCHOS**

```python
class Biblioteca:

    def __init__(self, nombre, ubicacion):
      self.nombre = nombre
      self.ubicacion = ubicacion
      self.libros = []  # Aquí está la relación uno a muchos

    def agregar_libro(self, libro):
        self.libros.append(libro)

    def mostrar_libros(self):
        return f"Libros en {self.nombre}:\n" + "\n".join(str(libro) for libro in self.libros)

    def __repr__(self):
      return f"{self.nombre} está ubicada en {self.ubicacion}"
```

```python
class Libro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor

    def __repr__(self):
        return f"'{self.titulo}' por {self.autor}"
```

```python
# Crear una biblioteca
mi_biblioteca = Biblioteca("Biblioteca Central", "Calle Desengaño 21")

# Crear algunos libros
libro1 = Libro("Cien años de soledad", "Gabriel García Márquez")
libro2 = Libro("Don Quijote de la Mancha", "Miguel de Cervantes")

# Agregar los libros a la biblioteca
mi_biblioteca.agregar_libro(libro1)
mi_biblioteca.agregar_libro(libro2)

# Mostrar información de la biblioteca y sus libros
print(mi_biblioteca)
print(mi_biblioteca.mostrar_libros())
```

**HERENCIA Y HERENCIA MÚLTIPLE**

Si queremos que la biblioteca contenga varios tipos de recursos, no solo de la clase Libro, que compartan algunas características en común, la Herencia puede ser útil.

```python
class MaterialBiblioteca:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor

    def __repr__(self):
        return f"{self.titulo} por {self.autor}"

    def ver_titulo(self):
        return f"Título: {self.titulo}"
```

```python
class Libro(MaterialBiblioteca):
    def __init__(self, titulo, autor, genero):
        super().__init__(titulo, autor)  # Constructor de la superclase
        self.genero = genero # Atributo propio de la subclase

    def info_completa(self):
        return f"{super().__repr__()} | Género: {self.genero}"
```

```python
libro = Libro("1984", "George Orwell", "Distopía")
print(libro)
print(libro.info_completa())
```

Herencia múltiple:

```python
class Digital:
    def __init__(self, formato):
        self.formato = formato

    def info_digital(self):
        return f"Disponible en formato: {self.formato}"
```

```python
class LibroDigital(Libro, Digital):
    def __init__(self, titulo, autor, genero, formato):
        Libro.__init__(self, titulo, autor, genero)
        Digital.__init__(self, formato)

    def info_completa(self):
        return f"{super().info_completa()} | {super().info_digital()}" #También funciona self.info_digital()
```

```python
ebook = LibroDigital("Neuromante", "William Gibson", "Ciencia Ficción", "PDF")
print(ebook.info_completa())
```

```python
print(LibroDigital.__mro__)
```

IMPORTANTE: En **herencia múltiple**, importa muchísimo el orden en el que se indican los múltiples padres de la subclase, sobretodo cuando dos o más superclases tienen algún método con el mismo nombre.
Esto va relacionado con el MRO: Method resolution orden.
