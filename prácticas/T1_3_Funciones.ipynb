## Apuntes y ejemplos prácticos sobre funciones

Las funciones se declaran usando la palabra reservada **def**, y suelen cerrarse con **return** para devolver un resultado.

Puede haber uno o varios **return** en el cuerpo de la función: esto dependerá de las estructuras de control que pueda haber dentro de la misma.

```python
def saludar(nombre):
    # Una función con un único return
    return "¡Hola, " + nombre + "!"

def evaluar_temperatura(grados):
    # Una función con múltiples return dependiendo de la condición
    if grados > 30:
        return "Hace mucho calor"
    else:
        return "La temperatura es agradable"

print(saludar("Mundo"))
print(evaluar_temperatura(35))
```

Si Python llega al final de la función sin encontrarse ninguna sentencia **return**, automáticamente devolverá un objeto `None`.

Una función puede tener dos tipos de argumentos:
*   **Posicionales**: su posición en la lista de argumentos ha de tenerse en cuenta al llamar a la función.
*   **De palabra clave**: se usan para especificar valores por defecto y argumentos que son opcionales (no es obligatorio pasarlos en la llamada a la función).

La principal restricción es que los argumentos de palabra clave deben ir después de los argumentos posiciones, si los hubiere.

```python
def funcion_vacia():
    # Esta función hace un cálculo pero no tiene 'return'
    suma = 2 + 2

resultado = funcion_vacia()
print("Lo que devuelve una función sin return es:", resultado)

print("-" * 20)

def presentar_usuario(nombre, edad, pais="España"):
    # 'nombre' y 'edad' son posicionales (obligatorios)
    # 'pais' es de palabra clave (opcional, tiene valor por defecto)
    print(nombre, "tiene", edad, "años y vive en", pais)

# Llamada normal usando el valor por defecto de 'pais'
presentar_usuario("Ana", 25)

# Llamada aportando un nuevo valor para 'pais'
presentar_usuario("Luis", 30, "México")
```

También puede tratarse a los argumentos posicionales como si fueran de palabra clave en la llamada a funciones, lo que da cierto grado de libertad a la hora de colocarlos dentro de los () en la llamada a la función.

```python
# Usando la función 'presentar_usuario' creada en la celda anterior.
# Al nombrar explícitamente los argumentos, el orden ya no importa:
presentar_usuario(edad=40, pais="Colombia", nombre="Carlos")
```

**Devolviendo múltiples resultados**

Python hace posible con una sintaxis intuitiva devolver varios resultados al finalizar una función.

```python
def obtener_maximo_y_minimo(lista_numeros):
    valor_max = max(lista_numeros)
    valor_min = min(lista_numeros)
    # Devolvemos los dos resultados separados por comas
    return valor_max, valor_min
```

En realidad, el *return* de la función no devuelve múltiples objetos, solo uno: una tupla, que después puede ser desempaquetada en una asignación múltiple donde interviene la llamada a la función.

```python
# Si guardamos el resultado en una sola variable, veremos que es una tupla
resultado = obtener_maximo_y_minimo([10, 2, 8, 4, 15])

print("El objeto devuelto es:", resultado)
print("El tipo de dato devuelto es:", type(resultado))
```

```python
# Podemos "desempaquetar" esa tupla directamente en dos variables:
mi_maximo, mi_minimo = obtener_maximo_y_minimo([10, 2, 8, 4, 15])

print("El valor máximo es:", mi_maximo)
print("El valor mínimo es:", mi_minimo)
```

El comportamiento flexible de las funciones como **objetos**. Vamos a ilustrarlo con un ejemplo de dos modos diferentes de aplicar un flujo de procesamiento de texto.

Supongamos que necesitamos aplicar una serie de procesamientos sobre un conjunto de cadenas de texto para hacer que tengan un formato uniforme.

```python
estados = ['    Alabama ', 'Georgia!', 'Georgia', 'georgia', 'FlOrida', 'south   carolina##', 'West virginia?']
```

Podemos utilizar funciones nativas de Python para el tratamiento de cadenas, así como la librería `re` para usar expresiones regulares:

```python
import re

def limpiar_textos(lista_cadenas):
    resultado = []
    for texto in lista_cadenas:
        texto = texto.strip()                 # 1. Quita espacios en los extremos
        texto = re.sub('[!#?]', '', texto)    # 2. Quita signos raros usando expresiones regulares
        texto = texto.title()                 # 3. Pone en mayúscula la primera letra de cada palabra
        resultado.append(texto)
    return resultado
```

```python
estados_limpios = limpiar_textos(estados)
print("Lista procesada de forma clásica:", estados_limpios)
```

Esta solución permite de forma muy flexible modificar el flujo de procesamiento dividido en etapas o *pipeline* que queramos aplicar sobre nuestros datos en un momento u otro.

```python
# Vamos a crear funciones muy pequeñas, cada una con un único propósito

def quitar_espacios(texto):
    return texto.strip()

def quitar_simbolos(texto):
    return re.sub('[!#?]', '', texto)

def capitalizar_letras(texto):
    return texto.title()
```

```python
# En Python, podemos agrupar funciones dentro de una lista (pipeline)
operaciones_a_realizar = [quitar_espacios, quitar_simbolos, capitalizar_letras]

def aplicar_pipeline(textos, operaciones):
    resultado = []
    for texto in textos:
        for funcion in operaciones:
            # Aplicamos secuencialmente cada función del pipeline al texto
            texto = funcion(texto)
        resultado.append(texto)
    return resultado
```

```python
estados_pipeline = aplicar_pipeline(estados, operaciones_a_realizar)
print("Lista procesada usando el pipeline de funciones:", estados_pipeline)
```

**Funciones anónimas (Lambda)**

Son funciones anónimas escritas de forma compacta, en una sola instrucción, cuyo resultado es el valor que devuelven. Se declaran con la palabra reservada `lambda`. No tienen nombre de función propio, de ahí que se les llame anónimas.

Se usan habitualmente para tareas puntuales de procesamiento sobre colecciones, con funciones de orden superior como `map()` y `filter()`.



```python
# Una función tradicional se vería así:
# def sumar_diez(x):
#     return x + 10

# Su equivalente en función lambda anónima es:
# lambda x: x + 10

# Podemos ejecutarla inmediatamente pasándole un valor, por ejemplo (5):
resultado_lambda = (lambda x: x + 10)(5)
print("El resultado de la lambda es:", resultado_lambda)
```

```python
# También pueden recibir múltiples argumentos
suma_anonima = (lambda a, b: a + b)(20, 30)
print("La suma usando lambda es:", suma_anonima)
```

Uso de lambda en **funciones de orden superior**:

```python
numeros = [1, 2, 3, 4, 5]

# map() aplica una función a todos los elementos de una colección.
# En lugar de definir una función aparte, pasamos una lambda directamente:
cuadrados = list(map(lambda x: x ** 2, numeros))

print("Números originales:", numeros)
print("Números al cuadrado:", cuadrados)
```

```python
# filter() filtra los elementos de una colección según una condición (devuelve True o False)
# Vamos a quedarnos solo con los números pares (cuyo resto de dividir entre 2 sea 0)
pares = list(filter(lambda x: x % 2 == 0, numeros))

print("Solo los números pares:", pares)
```

```python
words = ["elephant", "sun", "onion", "apple", "tree", "fish"]

# Ordena primero por longitud y, en caso de empate, alfabéticamente.
sorted_words = sorted(words, key=lambda word: (len(word), word))
print(sorted_words)
```

Python también permite guardar una función lambda en una variable, y después en otra parte del código llamarla con el nombre de la variable y el argumento que sea necesario.

```python
# Asignamos la función lambda a la variable 'es_mayor_de_edad'
es_mayor_de_edad = lambda edad: edad >= 18

# Ahora podemos usar esa variable como si fuera el nombre de una función normal
print("¿Alguien de 15 años es mayor?", es_mayor_de_edad(15))
print("¿Alguien de 21 años es mayor?", es_mayor_de_edad(21))
```

**Ventajas**: anónima, ligera e ideal cuando el cuerpo de la función es una sola expresión simple.
**Desventajas**: no permite bloques de código más complejos con estructuras de control.
