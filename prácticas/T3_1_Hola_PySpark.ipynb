```python
# !pip install pyspark

from pyspark.sql import SparkSession

# 1. Inicializar SparkSession
spark = SparkSession.builder \
    .appName("MiPrimeraApp") \
    .master("local[*]") \
    .getOrCreate()

# 2. Crear datos ficticios
data = [("Alice", 25), ("Bob", 30)]
cols = ["Nombre", "Edad"]

# 3. Crear y mostrar DataFrame
df = spark.createDataFrame(data, cols)
df.show()

# 4. Detener la sesión
spark.stop()
```

```python
# Ejemplo 2: Ejecución local restringida a un número específico de hilos/núcleos (ej. 2)
# Útil si no queremos que Spark consuma el 100% de nuestro procesador mientras hacemos otras cosas.
from pyspark.sql import SparkSession

spark_2 = SparkSession.builder \
    .appName("UsoRestringidoDeNucleos") \
    .master("local[2]") \
    .getOrCreate()

print(f"App Name: {spark_2.sparkContext.appName}")
print(f"Master: {spark_2.sparkContext.master}")
```

```python
# Ejemplo 3: Modificación de configuraciones internas de la sesión (ej. memoria)
# Si vamos a procesar datos más grandes en una máquina local, a veces necesitaremos subir la memoria del driver.
from pyspark.sql import SparkSession

# Indicar que podemos usar hasta 4GB de memoria RAM
# Reducir el número de particiones a 4 (por defecto es 200): útil si no vamos a manejar demasiados datos
spark_3 = SparkSession.builder \
    .appName("ConfiguracionDeMemoria") \
    .master("local[*]") \
    .config("spark.driver.memory", "4g") \
    .config("spark.sql.shuffle.partitions", "4") \
    .getOrCreate()

# Comprobamos si la configuración se aplicó correctamente
particiones = spark_3.conf.get("spark.sql.shuffle.partitions")
print(f"Particiones configuradas para shuffle: {particiones}")
```

```python
# Ejemplo 4: Habilitar soporte para Apache Hive
# Necesario si queremos interactuar con un metastore (almacén de datos) de Hive, guardar tablas persistentes
# o usar ciertas funciones SQL avanzadas.
from pyspark.sql import SparkSession

# Dotar nuestro SparkSession de la capacidad de leer y escribir código en un catálogo de Hive
spark_4 = SparkSession.builder \
    .appName("SoporteHiveHabilitado") \
    .master("local[*]") \
    .enableHiveSupport() \
    .getOrCreate()

print(f"App Name: {spark_4.sparkContext.appName} (con soporte Hive)")
```
