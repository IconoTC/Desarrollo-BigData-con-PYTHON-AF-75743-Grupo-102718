```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

# Inicialización de objeto SparkSession
spark = SparkSession.builder.appName("SparkSQLExample").getOrCreate()

# Creación de un dataset sencillo y pequeño
data = [("Alice", 25, "New York"),
        ("Bob", 30, "London"),
        ("Charlie", 20, "New York"),
        ("David", 17, "Paris"),
        ("Eve", 35, "London"),
        ("Frank", 19, "New York")]

# Definir esquema para el DataFrame: esencial cuando se crea a partir de datos 'en crudo'
schema = StructType([
    StructField("nombre", StringType(), True),
    StructField("edad", IntegerType(), True),
    StructField("ciudad", StringType(), True)
])

# Creación del DataFrame: datos en crudo + esquema
df = spark.createDataFrame(data, schema)

# Registramos el DataFrame como una vista temporal para poder utilizar Spark SQL
df.createOrReplaceTempView("usuarios")

print("SparkSession inicializado, DataFrame 'df' creado, y vista temporal 'usuarios' registrada.")
print("Sample data:")
df.show()
```

```python
# --- Enfoque 1: DataFrame API (Python, similar a Pandas) ---
df_resultado = df.filter(df.edad >= 18) \
                 .groupBy("ciudad") \
                 .count()

df_resultado.show()
```

```python
# --- Enfoque 2: SQL Nativo ---
df_resultado = spark.sql("""
    SELECT ciudad, COUNT(*) as count
    FROM usuarios
    WHERE edad >= 18
    GROUP BY ciudad
""")

df_resultado.show()
```
