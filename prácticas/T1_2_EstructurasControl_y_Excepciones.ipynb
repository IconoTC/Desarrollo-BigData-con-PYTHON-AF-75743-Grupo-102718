**Ejemplos con Estructuras de Control**

```python
temperatura = 25

if temperatura > 30:
  print("Hace calor, ¡tómate un helado!")
if temperatura > 20:
  print("Temperatura agradable, ¡tómate un zumo fresquito!")
if temperatura > 10:
  print("Hace fresco, ¿qué tal un café con leche?")
```

```python
# Versión más apropiada con un 'else' al final

temperatura = 25

if temperatura > 30:
  print("Hace calor, ¡tómate un helado!")
elif temperatura > 20:
  print("Temperatura agradable, ¡tómate un zumo fresquito!")
elif temperatura > 10:
  print("Hace fresco, ¿qué tal un café")
else:
  print("Hace frío, ¡un chocolate caliente!")
```

```python
# Otro ejemplo, con match-case

animal = "hamster"

match animal:
  case "perro":
    print("¡Los perros son cariñosos!")
  case "gato":
    print("¡Los gatos son independientes!")
  case "pájaro":
    print("¡Los pájaros son adorables!")
  case _:
    print("¡Ese animal suena interesante!")
```

```python
# Probando a implementar el ejemplo de la temperatura con un match-case

temperatura = 8

match temperatura:
  case 1, 2, 3:
    print("Hace calor")
  case 4, 5, 6:
    print("Clima agradable")
  case 7, 8, 9:
    print("Un poco fresco")
  case _:
    print("Hace mucho frío")
```

```python
num = int(input("Introduzca un número: "))

match num:
    case num if num > 0:
        print(f"{num} es un número positivo")
    case num if num < 0:
        print(f"{num} es un número negativo")
    case _:
       print("Ha introducido el número 0")
```

```python
temperatura = 7

match temperatura:
  case _ if temperatura > 30:
    print("Hace calor")
  case _ if temperatura > 20:
    print("Clima agradable")
  case _ if temperatura > 10:
    print("Un poco fresco")
  case _:
    print("Hace frío")
```

```python
# Condiciones compuestas con operadores lógicos

cocida = True
queso_derretido = False

if cocida and queso_derretido:
  print("¡La pizza está lista!")
elif cocida or queso_derretido:
  print("Casi lista ... ¡Un poco más de tiempo!")
else:
  print("Aún va para rato, siéntate tranquilamente")
```

Uso de condiciones compactas con el uso del **Operador Ternario**:

**<accion 1>** `if` **<condición>** `else` **<acción 2>**

```python
# Para realizar una asignación dinámica

edad = 19
mensaje = "Puede entrar al club" if edad >= 18 else "Lo siento, eres menor de edad"
print(mensaje)
```

```python
def saludo_matinal():
  return "¡Buenos días!"

def saludo_nocturno():
  return "¡Buenas noches!"

# Para llamar dinámicamente a una función u otra
hora = 19
print(saludo_matinal() if hora < 18 else saludo_nocturno())
```

```python
# Uso de argumentos dinámicos en llamadas a funciones
def calcular_precio(descuento):
  return 100 - descuento

es_vip = True
precio_final = calcular_precio(20 if es_vip else 5)
print(precio_final)
```

```python
# Generación de listas de forma dinámica

numeros = [x if x % 2 == 0 else "es impar" for x in range(1,6)]
print(numeros)
```

```python
# Devolver el resultado de una función

def evaluar_nota(nota):
  return "Aprobado" if nota >= 50 else "Suspenso"

print(evaluar_nota(75))
print(evaluar_nota(40))
```

**Formas de iterar con bucles en Python**

```python
# Iterar sobre una lista

for fruta in ["manzana", "banana", "cereza"]:
  print(fruta)
```

```python
frutas = ["manzana", "banana", "cereza"]
for i, fruta in enumerate(frutas):
  print(i, fruta)
```

```python
frutas = ["manzana", "banana", "cereza"]
for i in range(len(frutas)):
  print(i, frutas[i])
```

```python
nombres = {"Ana", "Juan", "Luis"}
for i, nombre in enumerate(nombres):
  print(i, nombre)

# No se recomienda iterar sobre Conjuntos (Sets), al no haber un orden establecido
```

```python
for i, letra in enumerate("Python"):
  print(i, letra)
```

```python
precios = {"manzana": 1.2, "banana": 0.5, "cereza": 2.0}

for i, clave in enumerate(precios):
  print(i, clave)
```

```python
for i, (clave, valor) in enumerate(precios.items()):
  print(i, clave, "cuesta", valor)
```

```python
# Iterar sobre dos colecciones a la vez con zip()

nombres = ["Ana", "Beatriz", "Carmen"]
edades = [25, 30, 22, 31, 28]

for nombre, edad in zip(nombres, edades):
  print(nombre, "tiene", edad, "años")
```

```python
# No es habitual, pero zip() puede usarse en diccionarios

dict1 = {"a": 1, "b": 2, "c": 3}
dict2 = {"a": "uno", "b": "dos", "c": "tres"}

for clave, palabra in zip(dict1, dict2):
  print(clave, "->", palabra)
```

```python
# Modificación que cobra más sentido

for num, palabra in zip(dict1.values(), dict2.values()):
  print(num, "->", palabra)
```

```python
for fruta in reversed(frutas):
  print(fruta)
```

```python
# Recorrer los pares/entradas de un diccionario

precios = {"manzana": 1.2, "banana": 0.5, "cereza": 2.0}

for fruta, precio in precios.items():
    print(fruta, "cuesta", precio)
```

```python
# Comprensión de listas

cuadrados = [x**2 for x in range(100,1,-10)]
print(cuadrados)
```

```python
# Bucle while con else, propio de Python

numeros = [1, 2, 3, 4, 5]
buscar = 6

while numeros:
    num = numeros.pop()
    if num == buscar:
        print(f"{buscar} encontrado.")
        break
else:
    print(f"{buscar} no fue encontrado.")
```

```python
# Ruptura de flujo de un bucle con 'continue/pass'

for i in range(1, 6):
  if i == 3:
    continue
  print(i)
```

**Manejo de excepciones**

```python
# Manejo de excepciones mediante condición

try:
  num = int(input("Introduce un número: "))
  print(10 / num)
except ZeroDivisionError:
  print("!No puedes dividir entre cero!")
except ValueError:
  print("Ingresa un número válido")
```

```python
# Manejo de excepciones con condición y else.
# La parte del else contendrá algo que solo queremos que se ejecute
# si todo el bloque try se ha ejecutado sin problemas.

try:
    num = int(input("Introduce un número: "))
    resultado = 10 / num
    print(f"El resultado es: {resultado}")
except ZeroDivisionError:
    print("¡No puedes dividir entre cero!")
except ValueError:
    print("Ingresa un número válido.")
else:
    print(f"El resultado es: {resultado}")
```

```python
try:
    archivo = open("archivo.txt", "r")
    contenido = archivo.read()
    print(contenido)
except FileNotFoundError:
    print("El archivo no se encontró.")
finally:
    print("El bloque `finally` siempre se ejecuta.")
    # Cerrar el archivo si se abrió correctamente
    try:
        archivo.close()
    except NameError:
        pass  # Si no se abrió el archivo, no intentamos cerrarlo
```
