## El tipo datetime

Python ofrece además un tipo especial para manejar fechas y horas. El principal tipo de *datetime* pero también están los tipos *date* y *time*.

Usar estos tipos requiere importarlos desde el módulo *datetime*

```python
from datetime import datetime, date, time

dt = datetime(2025, 3, 28, 20, 30, 21)
dt.month
```

```python
# Se pueden extraer los objetos date y time por separado

fecha = dt.date()
hora = dt.time()

print(f"Fecha: {fecha}. Hora: {hora}")
```

```python
# Podemos crear una cadena con el formato deseado de fecha-hora con esta función

dt.strftime('%d/%m/%Y %H:%M')
```

```python
# También se puede pasar una cadena a un objeto datetime:

nueva_fecha = datetime.strptime('20091031', '%Y%m%d')
nueva_fecha
```

```python
# Si deseamos eliminar información sobre minutos y segundos, podemos hacerlo así:
dt.replace(minute=0, second=0)
```

```python
# Python puede además inferir el día de la semana asociado a una fecha

dt.strftime('%w, %d/%m/%Y %H:%M')
```

También denominadas **tipos de datos complejos**, permiten agrupar en una misma variable varios elementos o valores normalmente interrelacionados, representando secuencias, conjuntos, o mapas.

## 1. Listas

Las listas son colecciones **ordenadas y mutables** que **permiten almacenar elementos duplicados de cualquier tipo**.

```python
juegos = ["Zelda: Breath of the Wild", "Mario Kart 8 Deluxe",
          "Super Smash Bros. Ultimate", "Pokémon Scarlet & Violet"]
print(juegos)
```

```python
# Añadir un nuevo juego al inventario
juegos.append("Animal Crossing: New Horizons")
print(juegos)
```

```python
# Eliminar un juego agotado
juegos.remove("Super Smash Bros. Ultimate")
print(juegos)
```

```python
# Ordenar los juegos alfabéticamente
juegos.sort()
print(juegos)
```

```python
# Añadir un nuevo juego en la segunda posición de la lista
juegos.insert(1, "The Legend of Zelda: Tears of the Kingdom")
print(juegos)
```

Un operador particularmente interesante y útil en las listas es el llamado *slicing* (rebanadas). Permite extraer un rango específico de elementos de la lista, simultáneos o no.

Veamos algunos ejemplos

```python
# Obtener los tres primeros juegos
primeros_tres = juegos[:3]
print(f"Primeros tres juegos: {primeros_tres}")

# Obtener los juegos desde el segundo hasta el final
juegos_desde_segundo = juegos[1:]
print(f"Juegos desde el segundo: {juegos_desde_segundo}")

# Obtener los dos últimos juegos
ultimos_dos = juegos[-2:]
print(f"Últimos dos juegos: {ultimos_dos}")

# Obtener un rango específico de juegos (del segundo al cuarto, sin incluir el cuarto)
rango_juegos = juegos[1:3]
print(f"Rango de juegos: {rango_juegos}")

# Obtener cada dos juegos de la lista, empezando desde el primero
cada_dos_juegos = juegos[::2]
print(f"Cada dos juegos: {cada_dos_juegos}")

# Invertir el orden de la lista
juegos_invertidos = juegos[::-1]
print(f"Juegos invertidos: {juegos_invertidos}")
```

### Resumen métodos con listas

`append()`: añade un nuevo elemento al final de la lista.

`extend()`: añade un grupo de elementos (iterables) al final, por ejemplo una nueva lista.

`insert(indice, elemento)`: Inserta un elemento en una posición.


`remove(elemento)`: elimina la primera ocurrencia del elemento.


`pop([i])`: obtiene y elimina el elemento de la lista en la posición i. Si no se especifica, obtiene y elimina el último elemento.


`clear()`: borra todos los elementos.

`index(elemento)`: obtiene el índice de la primera ocurrencia del elemento. Si no se encuentra, muestra `ValueError`.

`count(elemento)`: devuelve el número de ocurrencias del elemento.

`sort()`: ordena los elementos de la lista utilizando el operador.

`reverse()`: obtiene los elementos de la lista en orden inverso.

`copy()`: devuelve una copia poco profunda de la lista.

## 2. Tuplas

Las tuplas son colecciones **ordenadas e inmutables**, ideales para **proteger datos** que no deben cambiar durante el programa.

```python
precios = (
    ("Zelda: Breath of the Wild", 59.99),
    ("Mario Kart 8 Deluxe", 49.99),
    ("Super Smash Bros. Ultimate", 54.99),
    ("Pokémon Scarlet & Violet", 59.99)
)

# Mostrar los precios de los juegos
for juego, precio in precios:
    print(f"{juego}: ${precio}")
```

```python
# Intentando modificar el precio de un juego
# precios[0] = ("Zelda: Breath of the Wild", 54.99)
```

```python
# Desempaquetado de una tupla en múltiples variables
juego, precio = ("Zelda: Breath of the Wild", 59.99)
print(juego)
print(precio)
```

```python
# Si solo queremos un elemento de la tupla, la asignación se hace así:
nombre, _ = ("Mario Kart 8 Deluxe", 49.99)
print(nombre)
```

Las tuplas tienen **dos métodos propios** que las listas no tienen: *count(valor)*, *index(valor)*

```python
numeros = (1, 2, 3, 1, 1, 4)
print(numeros.count(1))
print(numeros.index(3))
```

Al igual que las listas, las tuplas también aceptan *slicing*

## 3. Conjuntos (set)

Los sets son **colecciones no ordenadas y mutables de elementos únicos**, ideales para eliminar duplicados y realizar operaciones matemáticas de conjuntos.


```python
clientes_vip = {"Ana", "Carlos", "Luis", "Sofía"}
clientes_vip
```

```python
# Un nuevo cliente VIP se une
clientes_vip.add("Miguel")
print(clientes_vip)
```

```python
# Un cliente deja de ser VIP
clientes_vip.remove("Carlos")
print(clientes_vip)
```

```python
# Verificar si un cliente es VIP
print("Ana" in clientes_vip)
print("Pedro" in clientes_vip)
```

## 4. Diccionarios (dict)

Los diccionarios son **colecciones de pares clave-valor mutables**, diseñadas para** buscar información de forma muy rápida** mediante claves únicas.


```python
clientes = {
    "Ana": {"edad": 30, "juegos_comprados": ["Zelda: Breath of the Wild", "Mario Kart 8 Deluxe"]},
    "Luis": {"edad": 25, "juegos_comprados": ["Super Smash Bros. Ultimate"]},
}

# Acceder a la información de un cliente
print(clientes["Ana"]["juegos_comprados"])
```

```python
# Agregar un nuevo cliente con su compra
clientes["Sofía"] = {"edad": 28, "juegos_comprados": ["Pokémon Scarlet & Violet"]}
print(clientes)
```

```python
# Un pequeño truco para imprimir el diccionario de forma más legible (JSON)
import json
print(json.dumps(clientes, indent=4))
```

```python
# Modificar la lista de juegos de un cliente
clientes["Ana"]["juegos_comprados"].append("Animal Crossing: New Horizons")
print(clientes["Ana"]["juegos_comprados"])
```
